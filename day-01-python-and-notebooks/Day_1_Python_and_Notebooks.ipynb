## Topic 2: Name Canteen Values and Spot Each Kind
**One-liner:** A variable is a named box that stores one value (`=` stores the right side under the left name), and every value has a data type you can check with `type()`.

### Variables: run order matters

```python
# Right order: define first, use second
total = 90
print(total)   # 90

# Wrong order would be: print(total) BEFORE total = 90 -> NameError.
# Restarting the notebook clears all variables, so re-run cells top to bottom.
```

### Variables: updating values

```python
plate_count = 2
unit_price = 45
print(plate_count * unit_price)   # 90

plate_count = 3                   # a new = REPLACES the old value
print(plate_count * unit_price)   # 135 (re-run the cells that use it)

plate_count = plate_count + 1     # a variable can use itself
print(plate_count)                # 4
```

### Naming variables the right way

```python
# Valid names: lowercase words joined by underscores
plate_count = 2
unit_price = 45
tea_cups = 3
total_1 = 90

# Invalid names (each would raise a SyntaxError, so they are left as comments):
# plate count = 2   -> spaces not allowed
# 2nd_plate = 1     -> cannot start with a digit
# tea-cups = 3      -> hyphen is read as minus
# class = 5         -> reserved word

# Names are case sensitive: price and Price are DIFFERENT variables
price = 10
Price = 20
print(price, Price)   # 10 20
```

### Basic data types and `type()`

```python
print(type(2))              # int   -> whole numbers
print(type(45.5))           # float -> numbers with decimals
print(type("Masala Dosa"))  # str   -> text inside quotes
print(type(True))           # bool  -> True or False
```

### Extending it: Digits still inside quotes

## Topic 3: Build Totals and Compare to a Limit
**One-liner:** Expressions combine values with operators: arithmetic (`+ - * / // % **`) gives numbers, comparison (`> < >= <= == !=`) gives `True`/`False`.

### Arithmetic operators

```python
print(45 + 55)    # 100  add
print(100 - 30)   # 70   subtract
print(45 * 2)     # 90   multiply
print(90 / 4)     # 22.5 divide (ALWAYS a float)
print(90 // 4)    # 22   floor divide (drops the decimal part)
print(90 % 4)     # 2    remainder
print(2 ** 3)     # 8    power
print(90 / 3)     # 30.0 / gives a float even when it divides evenly
```

### Building expressions and order of operations

```python
plate_count = 2
unit_price = 45
dosa_total = plate_count * unit_price   # an expression gives a result
print(dosa_total)                       # 90

# Order: ( ) first, then **, then * / // %, then + -
print(2 + 3 * 4)     # 14 (multiply first)
print((2 + 3) * 4)   # 20 (brackets first)
```

### Comparing values

```python
tea_total = 3 * 12   # 36 rupees

print(tea_total > 30)    # True
print(tea_total < 30)    # False
print(tea_total >= 36)   # True
print(tea_total <= 30)   # False
print(tea_total == 36)   # True
print(tea_total != 30)   # True

# One = STORES a value. Two == ASK a question and give a bool.
```

### Example: Dosa line total

```python
dosa_plates = 2
dosa_price = 45
dosa_line_total = dosa_plates * dosa_price   # * multiplies plates by price
print(dosa_line_total)                       # 90
```

### Add rice and compare tea

```python
# Cell 1: rice
rice_plates = 1
rice_price = 55
rice_line_total = rice_plates * rice_price
bill_so_far = dosa_line_total + rice_line_total   # reuses the dosa total from above
print(bill_so_far)                                # 145

# Cell 2: tea
tea_cups = 3
tea_price = 12
tea_line_total = tea_cups * tea_price
print(tea_line_total)         # 36
print(tea_line_total > 30)    # True -> tea crosses Rs 30
```

## Topic 4: Convert Quoted Cup Count and Check the Printout
**One-liner:** Digits inside quotes are text, so convert them with `int()` before any price math, then `print` every row to check it matches the board.

### Quotes matter: text vs number

```python
cup_count = "3"
print(cup_count * 12)    # '333333333333' -> text is REPEATED, not multiplied

cup_count = int("3")     # convert text -> whole number
print(cup_count * 12)    # 36
```

### Example: Fix tea cups, then multiply

```python
tea_cups_text = "3"               # still a str
tea_cups = int(tea_cups_text)     # int() builds a whole number
tea_price = 12
tea_line_total = tea_cups * tea_price
print(tea_cups, tea_line_total)   # 3 36
```

### Extending it: Adding text to a number

```python
# "3" + 12 raises TypeError: Python will not mix str and int with +

broken_total = "3" + 12

print(broken_total)

```

```python

# Fix: convert first, then do the math
print(int("3") + 12)   # 15
```

### One print line for the whole order

```python
# Reuses dosa_line_total, rice_line_total, tea_line_total from earlier cells
print("Dosa", dosa_line_total, "Rice", rice_line_total, "Tea", tea_line_total)
# Dosa 90 Rice 55 Tea 36
```

### Quiz 4
**Cup count is saved as `cup_count = "3"`. What is the safe next step before `cup_count * 12`?**

- A. `cup_count = int(cup_count)`
- B. `cup_count = cup_count + 12`
- C. `cup_count = "3" * 12`
- D. `cup_count = print(cup_count)`

**Answer: A.** `int()` turns quoted digits into a whole number. B fails (str + int), C repeats text, and D only prints (it would store `None`).

```python
cup_count = "3"
cup_count = int(cup_count)   # option A
print(cup_count * 12)        # 36
```
